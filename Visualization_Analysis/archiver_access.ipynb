### specify scan_id or uid

```python
user = 'leheny'
cycle = '2024_1'

scan = [147450]


##########################################################################################
uid_list = []; scan_list=[]
for s in scan:
    if type(s) == int:
        scan_list.append(s)
        uid=[h.start['uid'] for h in db(scan_id=s, user=user,cycle=cycle)][0]
        uid_list.append(uid)
    else:       
        h=db[s];uid_list.append(h.start['uid']);scan_list.append(h.start['scan_id'])
    h=db[uid_list[-1]]
    print('scan: %s -> uid: %s -> epoch [s]: %s'%(scan_list[-1],uid_list[-1],h.start['time']))
```
