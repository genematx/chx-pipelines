```python
# Note: because of the way bg is passed to the fit function as a static 'parameter', the function needs to be defined here...
def m_Voigt(x,a,b,c,d,e,f,g,bg):
    y=np.zeros(len(x))
    for i in range(len(a)):
        if i == 0:
            y+=pseudo_Voigt(x,a[i],b[i],c[i],d[i],e[i],f[i])
        else:
             y+=pseudo_Voigt(x,a[i],b[i],c[i],d[i],e=0,f=0)
    y+=g[0]*bg
    return y
    

def mm_Voigt(x,*C):
    s=int(len(C)/7) # this is the number of peaks
    a=C[0:s];b=C[s:2*s];c=C[2*s:3*s];d=C[3*s:4*s];e=C[4*s:5*s];f=C[5*s:6*s];g=C[6*s:7*s]

    return m_Voigt(x,a,b,c,d,e,f,g,bg)

# Note: functions below use global variables and need to be defined here
def sumOfSquaredError(parameterTuple):
    warnings.filterwarnings("ignore") # do not print warnings by genetic algorithm
    val =mm_Voigt(t_int_x[qind], *parameterTuple)
    return np.sum((y_data_est - val) ** 2.0)

def generate_Initial_Parameters(method = 'dual_annealing',workers=1,x0=None):
    if method == 'differential_evolution':
        result = differential_evolution(sumOfSquaredError, Bounds(bounds_[0],bounds_[1]), seed=3,vectorized=False,workers=workers,tol=.01,x0=x0,strategy='best1bin')
    elif method == 'dual_annealing':
        result = dual_annealing(sumOfSquaredError, Bounds(bounds_[0],bounds_[1]), seed=3, x0=x0)
    elif method == 'shgo':
        print('Warning: global optimization method "shgo" might be slow...')
        result = shgo(sumOfSquaredError, Bounds(bounds_[0],bounds_[1]), workers=workers)
    elif method == 'direct':
        print('Warning: global optimization method "direct" might be slow...')
        result = direct(sumOfSquaredError, Bounds(bounds_[0],bounds_[1]))
    else:
        raise Exception('method %s is unknown...method should be one of "differential_evolution", "dual_annealing", "shgo" or "direct".'%method)
    return result.x
```

```python
early_stop=0 # number of dataframes to skip at the end
in_situ_r2_thresh = .99 # if dataset is in-situ, will re-calculate starting values by global optimization if fit R2 is < in_situ_r2_thresh
in_situ_repeat = True # if dataset is in-situ, will re-do pattern fit, after calculating new starting values by global optimization
first_frame=0
if in_situ_crystallization:
    first_frame=xc+xf+x_crystal-crystallization_onset_dict['x_crystal_err']

if in_situ_crystallization or all_crystalline:
    if in_situ_crystallization: amplitude_scale=.1 # for in-situ crustallization need to start with small peak amplitudes
    else: amplitude_scale=1
    
    # limit fitting range
    q_start=9.;q_end=16.2
    ind1=t_int_x>=q_start;ind2=t_int_x<=q_end;ind=ind1*ind2
    bg=y_norm[ind]
    
    # get sensible starting values and bounds for fit
    starting_values_,bounds_,pn=waxs_fit_setup(material,amplitude_scale=amplitude_scale)    
    
    last_successful =  starting_values_
    peak_fit_dict={}
    tr=time.time()
    progress_bar = tqdm(total=len(range(first_frame,len(t_int_y)-early_stop)),desc='fitting progress',unit='frames',nrows=20)
    display(clear=True,wait =True)
    
    sigma_=None;starting_values
    _last_frame_= len(t_int_y)-early_stop
    roll_ave_thresh=10 # if the rolling average  (of the last 5 fitting times) is larger than this threshold AND we didn't do global optimzation of the starting values during the past 25 iterations, we will take the time to do one
    r2_roll_ave_thresh=.98 # if the rolling average  (of the last 5 fitting times) is smaller than this threshold and we didn't do global optimization during the past 10 iterations, we will take time to do one
    ft=[];roll_ave=[];rr2=[];r2_roll_ave=[]

    for ffrn,frame_nr in enumerate(range(first_frame,_last_frame_)):    
        if int_ind[frame_nr]: # make sure this is not a zero intensity frame, e.g. because the shutter was closed
            if all_crystalline: # create linear background between first and last peak in list
                bg=t_int_x[ind]**(2) # simply add a quadratic term to the background built into the pseudo-Voigt model
            if ffrn == 0:
                ### get starting  value from global optimization
                y_data_est=t_int_y[frame_nr][ind]
                print('Getting initial starting values by a global optimization method. This might take a minute...')
                t1=time.time()
                starting_values = generate_Initial_Parameters(x0=starting_values_)
                print('elapsed time for finding globally optimized starting values: %.3fs'%(time.time()-t1))
                starting_values_recalculated = [ffrn]    
            try: del popt # if fit fails (e.g. didn't converge) it will not return popt and plot will show parameters of last successful fit
            except: pass
            try:
                t3=time.time()
                ###################################
                popt, pcov = curve_fit(mm_Voigt,t_int_x[ind],t_int_y[frame_nr][ind]  , p0= starting_values,  bounds=bounds_,ftol=1E-6,xtol=1E-8,method='dogbox',sigma=sigma_)
                ###################################
                ft.append(time.time()-t3)
                r2=np.round(R_2(t_int_y[frame_nr][ind],mm_Voigt(t_int_x[ind], *popt)),4)
                rr2.append(r2)
                perr = np.sqrt(np.diag(pcov))
                sperr = split_array(7,perr)
                spopt=split_array(7,popt)

                recalculate_starting_values = False
                repeat_fit = False
                if len(ft)>5:    
                    roll_ave.append(np.mean(ft[-5:]))
                    if roll_ave[-1] > roll_ave_thresh and ffrn - starting_values_recalculated[-1] >= 25 and _last_frame_-ffrn > 5: # NEED TO MAKE SURE WE don't enter this if we're close to the last iteration anyways
                        recalculate_starting_values=True
                if len(rr2)>5:
                    r2_roll_ave.append(np.mean(rr2[-5:]))
                    if r2_roll_ave[-1] < r2_roll_ave_thresh and ffrn - starting_values_recalculated[-1] >= 10:
                         recalculate_starting_values=True
                    elif sum(np.array(rr2[-5:])==-1)>2:
                        recalculate_starting_values=True
                ## need something more strict to not miss all the crystallization kinetics in in-situ crystallization datasets!
                if in_situ_crystallization:
                    if r2 < in_situ_r2_thresh:
                        recalculate_starting_values = True
                        repeat_fit = True
                                
                if recalculate_starting_values:
                    print('iteration # %s: Re-calculating starting values using global optimization. This might take a minute...'%ffrn)
                    y_data_est=t_int_y[frame_nr][ind] # -> this could be the next one, IF we knew what's the end of the loop..., but we'd also need to know it's not going to be an empty frame...re-calculate for the current frame for now
                    starting_values = generate_Initial_Parameters(x0=starting_values_)
                    starting_values_recalculated.append(ffrn)
                
                ## repeat fit with new starting values, if requested:
                if in_situ_crystallization and repeat_fit:
                    t3=time.time()
                    ###################################
                    popt, pcov = curve_fit(mm_Voigt,t_int_x[ind],t_int_y[frame_nr][ind]  , p0= starting_values,  bounds=bounds_,ftol=1E-6,xtol=1E-8,method='dogbox',sigma=sigma_)
                    ###################################
                    ft[-1] = time.time()-t3 #replace value from first fit attempt
                    r2=np.round(R_2(t_int_y[frame_nr][ind],mm_Voigt(t_int_x[ind], *popt)),4)
                    rr2[-1] = r2 #replace value from first fit attempt
                    perr = np.sqrt(np.diag(pcov))
                    sperr = split_array(7,perr)
                    spopt=split_array(7,popt)
               
            except: sperr = np.nan*np.ones(np.shape(split_array(7,starting_values_)));spopt =np.nan*np.ones(np.shape(split_array(7,starting_values_)));r2=-1;rr2.append(r2) # in case the fit fails
        else: sperr = np.nan*np.ones(np.shape(split_array(7,starting_values_)));spopt = np.nan*np.ones(np.shape(split_array(7,starting_values_)));r2=-1 # in case intensity in the frame is 0

        peak_fit_dict[frame_nr]={'time':(frame_nr-xc)*frame_time,'q_start':q_start,'q_end':q_end,'fit_index':ind,'Q':t_int_x[ind],'bg':bg,'peak_num':len(spopt[0]),'R2':r2,'a':spopt[0],'b':spopt[1],'c':spopt[2],'d':spopt[3],'e':spopt[4],
                                 'f':spopt[5],'g':spopt[6],'param_definitions':{'a':'center','b':'amplitude','c':'fwhm','d':'mixing','e':'constant offset',
                                                                               'f':'slope','g':'background scale'},
                                'a_err':sperr[0],'b_err':sperr[1],'c_err':sperr[2],'d_err':sperr[3],'e_err':sperr[4],'f_err':sperr[5],'g_err':sperr[6]}
        if plotting and int_ind[frame_nr]:
            display(clear=True,wait =True)
            plt.title('time-resolved WAXS, scan_id: %s'%scan_list[0]+'\n%s'%exp_str)
            plt.plot(t_int_x[ind],t_int_y[frame_nr][ind],'k.',markersize=.8)
            try: # this will fail if the fit failed
                plt.plot(t_int_x[ind],mm_Voigt(t_int_x[ind],*popt),'r-',alpha=.8,label='#%s: %ss\nR$^2$=%s'%(frame_nr,np.round((frame_nr-xc)*frame_time,2),r2))
                plt.legend();plt.ylabel('intensity [arb.u.]');plt.xlabel('Q [nm$^{-1}$]');plt.grid(True)
                ### peaks without background
                sa = split_array(7,popt)
                for i in range(4,7):
                    sa[i]=np.zeros(len(sa[i]))
                popt_=np.array(sa).flatten().tolist()
                plt.plot(t_int_x[ind],mm_Voigt(t_int_x[ind],*popt_),'g--',linewidth=.7)
                sa_=split_array(7,popt_)
                for i in range(pn):
                    plt.plot([sa_[0][i],sa_[0][i]],[0,sa_[1][i]],'k--',linewidth=.7)
            except: pass
            display(plt.gcf())
            plt.close()
        progress_bar.update()
    print('total elapsed time for fitting: %.1fs'%(time.time()-tr))
else: peak_fit_dict=None
```
